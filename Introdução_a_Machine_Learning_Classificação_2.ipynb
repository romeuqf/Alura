```python
import pandas as pd
```

```python
uri = "https://gist.githubusercontent.com/guilhermesilveira/2d2efa37d66b6c84a722ea627a897ced/raw/10968b997d885cbded1c92938c7a9912ba41c615/tracking.csv"
mapa= {'home':'principal',
       'how_it_works':'como_funciona',
       'contact':'contato',
       'bought':'comprou'}
base = pd.read_csv(uri)
base.rename(columns=mapa,inplace=True)
base.head()
```

```python
x = base[['principal','como_funciona','contato']]
x.head()
```

```python
y = base['comprou']
y.head()
```

```python
base.shape
```

```python
treino_x = x[:75]
treino_y = y[:75]
teste_x = x[75:]
teste_y = y[75:]
```

```python
print('Treinaremos com %d e testaremos com %d elementos' %(len(treino_x),len(teste_x)))
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score
```

```python
modelo = LinearSVC()
modelo.fit(treino_x,treino_y)
```

```python
previsoes = modelo.predict(teste_x)
previsoes
```

```python
acuracio = accuracy_score(teste_y,previsoes)*100
print('A acuracia foi %.2f%%' %acuracio)
```

Separando Treino e Teste com Biblioteca

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

SEED = 20

treino_x, teste_x, treino_y, teste_y = train_test_split(x,y,random_state=SEED, test_size=0.25)
print(treino_x.shape)
print(teste_x.shape)
print('Treinaremos com %d e testaremos com %d elementos' %(len(treino_x),len(teste_x)))
modelo = LinearSVC()
modelo.fit(treino_x,treino_y)
previsoes = modelo.predict(teste_x)
acuracio = accuracy_score(teste_y,previsoes)*100
print('A acuracia foi %.2f%%' %acuracio)
```

Definir proporcionalidade

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

SEED = 20

treino_x, teste_x, treino_y, teste_y = train_test_split(x,y,random_state=SEED, test_size=0.25, stratify=y)
print(treino_x.shape)
print(teste_x.shape)
print('Treinaremos com %d e testaremos com %d elementos' %(len(treino_x),len(teste_x)))
modelo = LinearSVC()
modelo.fit(treino_x,treino_y)
previsoes = modelo.predict(teste_x)
acuracio = accuracy_score(teste_y,previsoes)*100
print('A acuracia foi %.2f%%' %acuracio)
```
